### V de Cramér

```python
# Función para calcular V de Cramér
import pandas as pd
import numpy as np
from scipy.stats import chi2_contingency

def cramers_v(x, y):
    confusion = pd.crosstab(x, y)
    chi2 = chi2_contingency(confusion)[0]
    n = confusion.sum().sum()
    phi2 = chi2 / n
    r, k = confusion.shape
    phi2corr = max(0, phi2 - ((k-1)*(r-1))/(n-1))
    rcorr = r - ((r-1)**2)/(n-1)
    kcorr = k - ((k-1)**2)/(n-1)
    return np.sqrt(phi2corr / min((kcorr-1), (rcorr-1)))
```

```python
# Aplicar V de Cramér en variables relevantes

print(f"tipo_dispositivo vs region: V = {cramers_v(df['tipo_dispositivo'], df['region']):.3f}")
print(f"miembro_premium vs tipo_dispositivo: V = {cramers_v(df['miembro_premium'], df['tipo_dispositivo']):.3f}")
print(f"miembro_premium vs region: V = {cramers_v(df['miembro_premium'], df['region']):.3f}")
print(f"abandono vs tipo_dispositivo: V = {cramers_v(df['abandono'], df['tipo_dispositivo']):.3f}")
print(f"abandono vs region: V = {cramers_v(df['abandono'], df['region']):.3f}")
print(f"miembro_premium vs abandono: V = {cramers_v(df['miembro_premium'], df['abandono']):.3f}")
```

## Sección 6 - Limitaciones y próximos pasos

### **Limitaciones**
Correlación ≠ causalidad: Todos los hallazgos son asociativos. No podemos afirmar causalidad con análisis de correlación.Multicolinealidad extrema: compras_mes e ingreso_anual tienen r>0.90. Si se usan juntas en un modelo, inflan la varianza y distorsionan la interpretación.
Variables binarias sin poder predictivo: Punto-biserial <0.10 y V de Cramér <0.13 en todas las cruces. 
miembro_premium y abandono son independientes de las variables continuas y categóricas actuales del dataset, lo que limita su predicción.
Falta de variables clave: No tenemos satisfacción del cliente, NPS, tiempo como cliente, quejas, uso real de beneficios premium ni motivo de abandono.Datos transversales: Es una foto de un momento, no tenemos evolución temporal para ver si el abandono ocurre después de bajar visitas.

### **Próximos pasos** 

Probar segmentación adicional y enriquecer datos:
1. Modelo de churn con nuevas variables: Incluir antigüedad, tickets de soporte, interacción con beneficios premium y encuestas de salida. Probar Random Forest / XGBoost para predecir abandono.
2.  Test A/B de retención: Dado que V=0.120 entre premium y abandono es bajo pero es el más alto, lanzar experimento: ofrecer premium gratis 1 mes a grupo en riesgo y medir retención real vs grupo control.
3. Optimizar embudo de publicidad: Como gasto_publicidad correlaciona con visitas (r∼0.58) pero no con compras, hacer análisis de conversión y heatmaps para mejorar la tasa de visita → compra antes de escalar presupuesto.
